# list in data structures
1.list is the collection of elements enclosed by []
2.list is a mutable - editable
3.its allowed duplicate elements
4.each element sepearated by comma 


```python
#create list
names=["shiva","vikram","vikas","divya"]
ages=[10,20,30,40,10,20,15.8,25.9]
print(names)
print(ages)
print(type(names))
print(type(ages))
```

```python
#concatenation of lists 
a=[10,20,30,40,50]
b=[40,50,60,70,80]
result=a+b
print(result)
```

```python
#repeating list
a=[10,20,30,40,50]
print(a*20)
```

```python
#methods or functions available in list
print(dir(list))
```

```python
#description 
print(help(list))
```

```python
#add elements to list-append(),extend(),insert()
numbers=[10,20,30,40,50]
print(numbers)
numbers.append(100)
print(numbers)
```

```python
numbers=[10,20,30,40,50]
print(numbers)
numbers .extend([100,200,300])
print(numbers)
```

```python
numbers=[10,20,30,40,50]
print(numbers)
numbers.insert(1,100)
print(numbers)
```

```python
#delete elements from list -pop(),remove()
numbers=[10,20,30,40,50]
print(numbers)
numbers.remove(30)
print(numbers)
```

```python
#sorting techniques - sort(reverse=true or reverse-false)
a=[6,2,4,5,1,3]
print(a)
a.sort(reverse=True) # true -desc and false-asc
print(a)
```

# Tuples in data structures
1. tuple a collection of elements enclosed by ()
2. tuple is immutable -not editable
3. its allowed duplicate elements
4. each element seperated by ,

```python
#create tuple
data=(10,20,30,40,50,"shiva","vikram",100000.25,120000.52)
print(data)
print(type(data))
```

```python
#methods or functions in tuples
print(dir(tuple))
```

```python
a=(10,20,30,10,30,10,10,30,10)
print(a.count(10))
print(a.count(10+20))
```
