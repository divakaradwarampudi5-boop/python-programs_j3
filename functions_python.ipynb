````python
#user defined functions 
#write a program to print sum of 2 numbers using functions 
def sum_numbers():
    a=int(input("enter first number:"))
    b=int(input("enter second number:"))
    result=a+b
    print(result)
    return result
sum_numbers()
````

````python
#write a program to print add two strings using function
def add_strings():
    a=str(input("enter first string"))
    b=str(input("enter second string"))
    result=a+b
    print(result)
    return
add_strings()
````
