# file handling in python
1. file is a collection of information and stored in memory
2. files are two types 1.text files and 2. binary files
3. text files `,txt, .py, .html, .ipynb, .java...etc
4. binary files` .exe, .mp4, .mp3, .apk ...etc
   

# file operations
1. create files - w node
3. write files -w mode
4. read files- r mode
5. append files- a mode

```python
#write a program to create file using python script
create_file=open("profile.txt" ,"w")
print("successfully file is created and verify in home directiory")
create_file.close()
```

```python
#write data to file using python script
write_file=open("profile.txt","w")
message="welcome to TIA in learning python programming"
write_file.write(message)
print("successfully write data file, please open profile.txt file")
write_file.close()
```
