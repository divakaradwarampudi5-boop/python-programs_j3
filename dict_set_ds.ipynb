# dict in data structures
1.dict is the collection of items enclosed by {} - here item is a key:value pair 

```python
# create dict 
info={"name":"shiva","location":"hyderabad","salary":15000}
print(info)
print(type(info))
```

```python
# methods or functions
print(dir(dict))
```

```python
#keys , values, items,get
info={"name":"shiva","location":"hyderabad","salary":15000}
print(info)
print(info.keys())
print(info.values())
print(info.items())
print(info.get("location"))
```

```python
#update()
info={"name":"shiva","location":"hyderabad","salary":15000}
print(info)
info.update({"name":"vikram"})
print(info)
info.update({"domin":"data engineer"})
print(info)
```

```python
#pop()
info={"name":"shiva","location":"hyderabad","salary":15000}
print(info)
info.pop("location")
print(info)
```

# sets in data structures
1. set is collection of unorder unique elements enclosed by {}

```python
#create set
a={9,1,8,5,3,2,6,7,1,1,2,3,1}
print(a)
print(type(a))
```

```python
# methods or functions in sets
print(dir(set))
```

```python
#add()
a={10,20,50,40,30}
print(a)
a.add(100)
print(a)
```

```python
#difference
a={1,2,3,4,5}
b={3,4,5,6,7,8}
c=a.difference(b)
print(c)
```

```python
#union
a={1,2,3,4,5}
b={3,4,5,6,7,8}
c=a.union(b)
print(c)
```

```python
#intersection
a={1,2,3,4,5}
b={3,4,5,6,7,8}
c=a.intersection(b)
print(c)
```

```python
#discord
a={10,20,30,40,50,60}
print(a)
a.discord(40)
print(a)
```
