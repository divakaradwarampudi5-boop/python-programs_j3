# iterations in python
1. iterations means loops to work with repeatative taks or repeatative performances
2. python have two types of iterators
3. `for` loop
4. `while` loop

```python - for loop
for iter_variable in sequence:
    print("for is the keyword to start the for loop")
    print("iter_variable is one type of variable to iterate data from sequence")
    print("in is keyword to valiadate sequence is present or not")
    print("sequence is collection like range,list,string,tuple,...etc")
```

```python -while loop 
while (condition):
    print("while is keyword to start the iteration based on condition")
    print("condition is true then executed inside statements")
```

````python
#write a program to find the vowels from string using for loop 
name=str(input("please enter your name here:"))
for char in name:
    if (char in "aeiouAEIOU"):
        print(char)
````

````python
#write a program for non vowel from string using loop

text = input("Enter a string: ")
vowels = "aeiouAEIOU"
for char in text:
    if char not in vowels:
        print(char)
````
