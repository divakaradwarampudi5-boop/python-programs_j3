# Strings in data structures 
1. string is a collection of characters or enclosed by " " or ' '


```python
# create string 
name="shivaji"
info='welcome to TIA'
print(name)
print(info)
print(type(name))
print(type(info))
```

```python
# multi line string
info="""welcome to TIA
welcome to python programming 
Python with data analysis 
python with ML/AI
"""
print(info)
print(type(info))
```

```python
#concatenation of two or more strings
a="python"
b="program"
print(a+b)
```

```python
#repeate string
name="Machine Learning"
print(name)
print(name*10)
```

### accessing characters from string -Index based 
#### two ways 
1.Forward index - read characters from left to right (0,1,2,3,.....n-1)
2.Backward or nigitive index - read characters from right to left (-1,-2,-3,.....-n)


```python
name="Machine Learning"
print(name)
print(name[0])
    
```

```python
#slicing concept- accesing group of characters - string[start_index:end_index-1]
name="machine Learning"
print(name)
print(name[0:7])

```

```python
name="Machine Learning"
print(name)
print(name[::-1])
```
